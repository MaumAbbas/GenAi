```python
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage,SystemMessage
```

```python
model = init_chat_model(
    "openai/gpt-oss-20b",
    model_provider="groq"
)
```

```python
model = init_chat_model(
    "openai/gpt-oss-20b",
    model_provider="groq"
)
```

```python
#create message
messages=[
    SystemMessage("You are a helpful AI assistantat"),
    HumanMessage("What are the top 2 benifits  of using langchian?")
]
```

```python
#invoking the model 
response = model.invoke(messages)
response.content
```

```python
#streaming the answer 
for chunk in model.stream("Explain machine learning"):
    print(chunk.content, end="")
```

DYNAMIC PROMPT TEMPLATES

```python
from langchain_core.prompts import ChatPromptTemplate
translation_template = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a professional translator. "
        "Translate the following text {text} "
        "from {source_language} to {target_language}."
    ),
    (
        "user",
        "{text}"
    )
])

prompt = translation_template.invoke({
    "source_language": "English",
    "target_language": "Hinglish",
    "text": "LangChain makes building AI applications incredibly easy!"
})
```

```python
prompt
```

```python
translation_response=model.invoke(prompt)
print(translation_response.content)
```
